```python
print("Exact duplicate rows:", df.duplicated().sum())

print(
    "Duplicate cleaned texts:",
    df["cleaned_text"].duplicated().sum()
)
```

```python
duplicate_cleaned = df[df["cleaned_text"].duplicated(keep=False)]

duplicate_cleaned[
    ["tweet_text", "cleaned_text", "cyberbullying_type"]
].sort_values("cleaned_text").head(20)
```

```python
# Find cleaned texts that have more than one label
label_conflicts = (
    df.groupby("cleaned_text")["cyberbullying_type"]
      .nunique()
)

conflicting_texts = label_conflicts[label_conflicts > 1]

print("Cleaned texts with conflicting labels:", len(conflicting_texts))
```

```python
print("Total rows involved in conflicting labels:",
      df[df["cleaned_text"].isin(conflicting_texts.index)].shape[0])
```

```python
df[df["cleaned_text"].isin(conflicting_texts.index)][
    ["tweet_text", "cleaned_text", "cyberbullying_type"]
].sort_values("cleaned_text").head(30)
```

```python
exact_text_conflicts = (
    df.groupby("tweet_text")["cyberbullying_type"]
      .nunique()
)

exact_text_conflicts = exact_text_conflicts[
    exact_text_conflicts > 1
]

print(
    "Original tweets with conflicting labels:",
    len(exact_text_conflicts)
)
```

```python
print(
    "Rows involved:",
    df[df["tweet_text"].isin(exact_text_conflicts.index)].shape[0]
)
```

```python
# Rows with original tweet texts that have conflicting labels
conflict_rows = df[df["tweet_text"].isin(exact_text_conflicts.index)]

print("Rows to remove:", len(conflict_rows))
print("Rows remaining:", len(df) - len(conflict_rows))
```

```python
print(
    conflict_rows["cyberbullying_type"]
    .value_counts()
)
```

```python
# Remove rows where the same original tweet has conflicting labels
df_clean = df[
    ~df["tweet_text"].isin(exact_text_conflicts.index)
].copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
```

```python
df_clean = df_clean.drop_duplicates(
    subset=["tweet_text", "cyberbullying_type"]
).copy()

print("Rows after duplicate removal:", len(df_clean))
```

```python
print(
    df_clean["cyberbullying_type"]
    .value_counts()
)
```

```python
print(
    df_clean["cyberbullying_type"]
    .value_counts(normalize=True).round(4) * 100
)
```

```python
from pathlib import Path

output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
output_path = output_dir / "cyberbullying_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Saved to:", output_path)
print("Rows:", len(df_clean))
print("Columns:", df_clean.columns.tolist())
```
